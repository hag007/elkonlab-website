# Background

## Why rare variants
Common variants explain much of the heritability of complex traits, but each has a small effect and usually acts through regulation. Rare coding variants can have large effects and point directly to a gene. Exome and genome sequencing of hundreds of thousands of people in biobanks has made rare-variant association studies routine ([Backman et al. 2021](https://doi.org/10.1038/s41586-021-04103-z); [Karczewski et al. 2022](https://doi.org/10.1016/j.xgen.2022.100168)).

## Association tests
Single rare variants are rarely testable on their own, so they are grouped by gene or region ([Lee et al. 2014](https://doi.org/10.1016/j.ajhg.2014.06.009)):
- **Burden tests** sum the qualifying variants in a gene; they are powerful when most variants act in the same direction.
- **Variance-component tests**, such as [SKAT](https://doi.org/10.1016/j.ajhg.2011.05.029), allow effects in both directions. [SKAT-O](https://doi.org/10.1016/j.ajhg.2012.06.007) combines burden and SKAT, and [ACAT](https://doi.org/10.1016/j.ajhg.2019.01.002) combines p-values across tests and variant masks.
- **Biobank-scale tools** handle relatedness, unbalanced case-control ratios and many phenotypes: [SAIGE-GENE+](https://doi.org/10.1038/s41588-022-01178-w) and [REGENIE](https://doi.org/10.1038/s41588-021-00870-7). [GATE](https://doi.org/10.1038/s41467-022-32885-x) extends these tests to time-to-event outcomes.

## Annotation and meta-analysis
Choosing which variants to include matters as much as the test. [STAAR](https://doi.org/10.1038/s41588-020-0676-4) weights variants by several functional annotations, which also lets it test non-coding rare variants. Experimental scores from multiplexed assays ([MaveDB](https://doi.org/10.1186/s13059-019-1845-6)) add direct evidence for missense variants. Meta-analysis across cohorts uses [RAREMETAL](https://doi.org/10.1093/bioinformatics/btu367) and [Meta-SAIGE](https://doi.org/10.1038/s41588-025-02403-y).

## Heritability
[GREML-LDMS](https://doi.org/10.1038/ng.3390) estimates heritability by frequency and LD bins. Burden heritability shows that rare coding variants explain only a small part of heritability for most traits, but concentrate it in a few constrained genes ([Weiner et al. 2023](https://doi.org/10.1038/s41586-022-05684-z)). Rare variants can also modify the effect of a known pathogenic variant: damaging missense variants in innate immunity genes are associated with earlier breast cancer onset in BRCA1 185delAG carriers ([Shemesh et al. 2026](https://doi.org/10.1136/jmg-2025-111394)).
