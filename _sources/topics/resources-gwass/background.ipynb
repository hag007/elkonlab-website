# Background

## Large GWASs as resources
GWAS summary statistics are the input to most downstream methods: heritability and genetic correlation ([LD score regression](https://doi.org/10.1038/ng.3211)), fine-mapping, functional interpretation and polygenic scores. A good GWAS for method development is large, well powered and publicly available ([Uffelmann et al. 2021](https://doi.org/10.1038/s43586-021-00056-9)).

## Widely used GWASs
- **Breast cancer.** The BCAC meta-analysis of about 230,000 women ([Michailidou et al. 2017](https://doi.org/10.1038/nature24284)).
- **Prostate cancer.** A multi-ancestry meta-analysis with 269 risk variants ([Conti et al. 2021](https://doi.org/10.1038/s41588-020-00748-0)).
- **Height.** A saturated map of common variants from 5.4 million people ([Yengo et al. 2022](https://doi.org/10.1038/s41586-022-05275-y)).
- **Schizophrenia.** 287 loci from the Psychiatric Genomics Consortium ([Trubetskoy et al. 2022](https://doi.org/10.1038/s41586-022-04434-5)).
- **Type 2 diabetes.** Fine-mapping to single-variant resolution ([Mahajan et al. 2018](https://doi.org/10.1038/s41588-018-0241-6)).
- **Coronary artery disease.** Over a million participants ([Aragam et al. 2022](https://doi.org/10.1038/s41588-022-01233-6)).
- **Alzheimer's disease.** 75 risk loci ([Bellenguez et al. 2022](https://doi.org/10.1038/s41588-022-01024-z)).
- **Biobank meta-analyses.** The Global Biobank Meta-analysis Initiative combines biobanks across ancestries ([Zhou et al. 2022](https://doi.org/10.1016/j.xgen.2022.100192)).

## Catalogs
The [GWAS Catalog](https://doi.org/10.1073/pnas.0903103106) collects published associations and summary statistics ([Buniello et al. 2019](https://doi.org/10.1093/nar/gky1120)). [Open Targets Genetics](https://doi.org/10.1093/nar/gkaa840) adds fine-mapping and locus-to-gene scores. Atlases of results across thousands of traits show how widespread pleiotropy is ([Watanabe et al. 2019](https://doi.org/10.1038/s41588-019-0481-0)).
