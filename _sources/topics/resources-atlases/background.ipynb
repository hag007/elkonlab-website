# Background

## Why atlases
Interpreting a genetic association requires knowing where and when genes are expressed and regulated. Reference atlases supply this context for most of the methods in the other topics.

## Expression and molecular QTLs
- **[GTEx](https://doi.org/10.1038/ng.2653)** profiled gene expression and eQTLs in about 50 tissues from nearly 1,000 donors ([GTEx Consortium 2020](https://doi.org/10.1126/science.aaz1776)).
- **[eQTLGen](https://doi.org/10.1038/s41588-021-00913-z)** mapped cis- and trans-eQTLs in blood from over 30,000 people.
- **[Human Protein Atlas](https://doi.org/10.1126/science.1260419)** maps protein expression across tissues.

## Regulatory annotation
- **[ENCODE](https://doi.org/10.1038/nature11247)** annotated functional elements of the human genome; its third phase expanded this to millions of candidate regulatory elements ([ENCODE 2020](https://doi.org/10.1038/s41586-020-2493-4)).
- **[Roadmap Epigenomics](https://doi.org/10.1038/nature14248)** provided reference epigenomes for over 100 tissues and cell types.

## Single-cell atlases
- **[Human Cell Atlas](https://doi.org/10.7554/eLife.27041)** aims to map every cell type of the human body.
- **[Tabula Sapiens](https://doi.org/10.1126/science.abl4896)** profiled about 500,000 cells from 24 organs of the same donors.
- **[OneK1K](https://doi.org/10.1126/science.abf3041)** mapped single-cell eQTLs in blood from about 1,000 people.

## Points to keep in mind
- Disease states, development and stimulated conditions are underrepresented.
- Donor numbers are often small for single-cell atlases.
