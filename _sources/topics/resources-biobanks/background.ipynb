# Background

## What biobanks provide
Population biobanks link genotypes, and increasingly whole-exome or whole-genome sequences, to health records, questionnaires, imaging and biomarkers for hundreds of thousands of people. They are the basis of modern GWAS, rare-variant studies and risk-prediction research.

## Major biobanks
- **[UK Biobank](https://doi.org/10.1038/s41586-018-0579-z)** has about 500,000 participants aged 40–69 at recruitment ([Sudlow et al. 2015](https://doi.org/10.1371/journal.pmed.1001779)). Exome sequencing ([Backman et al. 2021](https://doi.org/10.1038/s41586-021-04103-z)) and whole-genome sequencing ([Halldorsson et al. 2022](https://doi.org/10.1038/s41586-022-04965-x)) are now available for nearly all participants.
- **[All of Us](https://doi.org/10.1056/NEJMsr1809937)** aims for one million diverse US participants, with an emphasis on groups underrepresented in research; its [genomic data](https://doi.org/10.1038/s41586-023-06957-x) include short-read genomes of over 200,000 participants.
- **[FinnGen](https://doi.org/10.1038/s41586-022-05473-8)** uses the Finnish population isolate, in which some rare deleterious variants are enriched.
- **[Biobank Japan](https://doi.org/10.1016/j.je.2016.12.005)** and the **[Million Veteran Program](https://doi.org/10.1016/j.jclinepi.2015.09.016)** add large East Asian and multi-ancestry cohorts.

## Reference panels
[1000 Genomes](https://doi.org/10.1038/nature15393) and [TOPMed](https://doi.org/10.1038/s41586-021-03205-y) provide reference haplotypes for imputation and LD estimation. [gnomAD](https://doi.org/10.1038/s41586-020-2308-7) gives allele frequencies and gene constraint metrics from about 140,000 people.

## Points to keep in mind
- Participants are usually healthier than the general population (selection bias).
- Most participants are of European ancestry, which limits portability (see *Polygenic risk prediction*).
