# Background

## Why interpretation is hard
Genome-wide association studies (GWAS) have mapped many thousands of loci for complex traits and diseases ([Visscher et al. 2017](https://doi.org/10.1016/j.ajhg.2017.06.005)), but about 90% of the associated variants lie outside protein-coding sequence. Risk variants are strongly enriched in regulatory DNA, such as DNase I hypersensitive sites, in the cell types relevant to each trait ([Maurano et al. 2012](https://doi.org/10.1126/science.1222794)). Associations are also spread across much of the genome, as described by the [omnigenic model](https://doi.org/10.1016/j.cell.2017.05.038). The main task is to go from a statistical signal to a biological mechanism: which tissues and cell types, which genes, and which pathways.

## Main approaches
- **Heritability enrichment.** [Stratified LD score regression (S-LDSC)](https://doi.org/10.1038/ng.3404) partitions SNP-heritability across functional annotations. [LDSC-SEG](https://doi.org/10.1038/s41588-018-0081-4) applies it to tissue- and cell-type-specific expressed genes to find the tissues that drive a trait.
- **Gene-level and gene-set analysis.** [MAGMA](https://doi.org/10.1371/journal.pcbi.1004219) aggregates SNP associations into gene and pathway scores. Assigning SNPs to genes by position alone can miss distal regulation, but adding regulatory interactions to the mapping is not a simple fix: in a controlled analysis it only occasionally strengthened gene-set results, and it needs controls for several pitfalls and confounders ([Groenewoud et al. 2022](https://doi.org/10.1371/journal.pcbi.1009908)).
- **Colocalization and molecular QTLs.** [coloc](https://doi.org/10.1371/journal.pgen.1004383) and [eCAVIAR](https://doi.org/10.1016/j.ajhg.2016.10.003) test whether a GWAS signal and an eQTL share a causal variant. Many GWAS hits still lack a matching eQTL, partly because eQTL and GWAS studies are biased toward different kinds of genes ([Mostafavi et al. 2023](https://doi.org/10.1038/s41588-023-01529-1)).
- **TWAS and PWAS.** [FUSION](https://doi.org/10.1038/ng.3506) and [S-PrediXcan](https://doi.org/10.1038/s41467-018-03621-1) impute expression from genotypes and test it for association; [PWAS](https://doi.org/10.1186/s13059-020-02089-x) does the same for predicted protein function. Their opportunities and pitfalls, such as co-regulated genes and LD, are reviewed by [Wainberg et al. 2019](https://doi.org/10.1038/s41588-019-0385-z).
- **Locus-to-gene models.** [L2G](https://doi.org/10.1038/s41588-021-00945-5) combines fine-mapping, distance, QTL and chromatin features into one causal-gene score.

## Deep-learning-based interpretation
Sequence-based models (see *Non-coding variant effect prediction*) score how each variant changes chromatin or expression in many cellular contexts. Aggregating these scores over the variants of a GWAS links the trait to specific cellular contexts, for example in [IMPACT-DL](https://doi.org/10.64898/2026.05.25.726449).

## Open questions
- How to bring in cell states and developmental stages that bulk eQTL atlases do not cover.
- How to combine many weak, correlated lines of evidence into calibrated gene and cell-type calls.
