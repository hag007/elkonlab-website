# Background

## What APA is
Most human genes have several polyadenylation sites. Alternative cleavage and polyadenylation (APA) produces mRNA isoforms with different 3′ ends, most often with longer or shorter 3′UTRs ([Elkon et al. 2013](https://doi.org/10.1038/nrg3482); [Tian & Manley 2017](https://doi.org/10.1038/nrm.2016.116)). Since the 3′UTR carries binding sites for microRNAs and RNA-binding proteins, APA changes mRNA stability, translation and localization.

## APA in proliferation and cancer
Proliferating cells express mRNAs with shorter 3′UTRs ([Sandberg et al. 2008](https://doi.org/10.1126/science.1155390)), and cancer cells use shortening to escape microRNA repression of oncogenes ([Mayr & Bartel 2009](https://doi.org/10.1016/j.cell.2009.06.016)). Proliferation is associated with global 3′UTR shortening and with more cleavage at intronic poly(A) sites; genes encoding 3′-end processing proteins are more highly expressed in proliferating cells, and E2F transcription factors contribute to this ([Elkon et al. 2012](https://doi.org/10.1186/gb-2012-13-7-r59)). Loss of the CFIm25 subunit (NUDT21) shortens 3′UTRs and increases tumour growth in glioblastoma ([Masamha et al. 2014](https://doi.org/10.1038/nature13261)). APA changes are also found during colorectal cancer development ([Morris et al. 2012](https://doi.org/10.1158/1078-0432.ccr-12-0543)).

## Regulators
The core machinery includes CPSF, CstF, CFIm and CFIIm. Other proteins tune site choice; for example, PABPN1 suppresses the use of proximal sites ([Jenal et al. 2012](https://doi.org/10.1016/j.cell.2012.03.022)). APA can also produce downstream uncapped RNA isoforms ([Malka et al. 2022](https://doi.org/10.1016/j.molcel.2022.09.036)).

## Mapping APA
3′-end sequencing methods, such as [PAS-seq](https://doi.org/10.1261/rna.2581711) and [QuantSeq](https://doi.org/10.1038/nmeth.f.376), measure polyA site usage directly. [DaPars](https://doi.org/10.1038/ncomms6274) and [QAPA](https://doi.org/10.1186/s13059-018-1414-4) infer APA from standard RNA-seq.

## 3′UTR regulation and genetics
MicroRNA targets are predicted by tools such as [TargetScan](https://doi.org/10.1016/S0092-8674%2803%2901018-3). RNA-binding proteins can control microRNA access, for example Pumilio on p27 ([Kedde et al. 2010](https://doi.org/10.1038/ncb2105)). Genetic variants can disrupt polyadenylation signals and change 3′UTR usage ([Shulman & Elkon 2020](https://doi.org/10.1371/journal.pgen.1008977)). 3′aQTL mapping shows that APA variation explains part of GWAS heritability ([Li et al. 2021](https://doi.org/10.1038/s41588-021-00864-5)).

## Open questions
- APA in single cells and in specific cell states.
- Which APA changes are causal and which are by-products of transcription changes.
