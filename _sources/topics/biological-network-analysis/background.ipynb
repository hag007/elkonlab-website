# Background

## Why networks
Disease genes do not act alone; they cluster in connected neighbourhoods of the molecular interaction network, called *disease modules* ([Barabási et al. 2011](https://doi.org/10.1038/nrg2918); [Menche et al. 2015](https://doi.org/10.1126/science.1257601)). Networks such as [STRING](https://doi.org/10.1093/nar/28.18.3442) and the reference interactome [HuRI](https://doi.org/10.1038/s41586-020-2188-x) let weak genetic signals be amplified by the signals of their neighbours.

## Network propagation
Propagation spreads scores from seed genes over the network, so that genes near many seeds score high ([Cowen et al. 2017](https://doi.org/10.1038/nrg.2017.38)). Early applications prioritized disease genes ([Vanunu et al. 2010](https://doi.org/10.1371/journal.pcbi.1000641)). [HotNet2](https://doi.org/10.1038/ng.3168) finds significantly mutated subnetworks in cancer. For GWAS, [NAGA](https://doi.org/10.1016/j.isci.2019.05.025) propagates gene-level association signals, and [NetWAS](https://doi.org/10.1038/ng.3259) reranks GWAS genes with tissue-specific networks.

## Module discovery
Active module identification searches for connected subnetworks enriched in a signal, such as differential expression. [DIAMOnD](https://doi.org/10.1371/journal.pcbi.1004120) grows modules from seed genes by connectivity significance. [WGCNA](https://doi.org/10.1186/1471-2105-9-559) builds co-expression modules instead of using interaction networks. Many module methods report modules that are not reproducible and that inflate GO enrichment. [DOMINO](https://doi.org/10.15252/msb.20209593) was designed to reduce such false calls, and it is also available as a [web server](https://doi.org/10.1093/bioinformatics/btac067).

## Open questions
- Interaction networks are biased toward well-studied genes.
- Tissue- and cell-type-specific networks are still incomplete.
