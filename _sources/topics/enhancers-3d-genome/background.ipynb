# Background

## Enhancers
Enhancers are short DNA elements, bound by transcription factors, that activate genes from a distance, often tens to hundreds of kilobases away ([Shlyueva et al. 2014](https://doi.org/10.1038/nrg3682)). Variants associated with common diseases are usually non-coding, so mapping regulatory elements and their functions is central to interpreting genetic risk ([Elkon & Agami 2017](https://doi.org/10.1038/nbt.3863)).

## Mapping enhancers
- **Chromatin profiling** marks candidate enhancers by open chromatin and histone modifications: [ChIP-seq](https://doi.org/10.1126/science.1141319), [ATAC-seq](https://doi.org/10.1038/nmeth.2688) and [CUT&Tag](https://doi.org/10.1038/s41467-019-09982-5).
- **Enhancer RNAs.** Active enhancers are transcribed, and these eRNAs are needed for p53-dependent enhancer activity ([Melo et al. 2013](https://doi.org/10.1016/j.molcel.2012.11.021)).
- **Functional assays** test enhancer activity directly: [MPRA](https://doi.org/10.1038/nbt.2137), [STARR-seq](https://doi.org/10.1126/science.1232542), CRISPR-Cas9 enhancer screens ([Korkmaz et al. 2016](https://doi.org/10.1038/nbt.3450)) and [CRISPRi screens](https://doi.org/10.1126/science.aag2445), up to genome-wide scale ([Gasperini et al. 2019](https://doi.org/10.1016/j.cell.2018.11.029)).

## 3D genome
Chromosome conformation capture shows how DNA folds in the nucleus. [Hi-C](https://doi.org/10.1126/science.1181369) revealed topologically associating domains (TADs; [Dixon et al. 2012](https://doi.org/10.1038/nature11082)) and chromatin loops anchored by CTCF ([Rao et al. 2014](https://doi.org/10.1016/j.cell.2014.11.021)). [Micro-C](https://doi.org/10.1016/j.cell.2015.05.048), [HiChIP](https://doi.org/10.1038/nmeth.3999) and [promoter capture Hi-C](https://doi.org/10.1038/ng.3286) increase resolution or focus on promoters. Disrupting TAD boundaries can rewire enhancer–promoter contacts and cause disease ([Lupiáñez et al. 2015](https://doi.org/10.1016/j.cell.2015.04.004)). A meta-analysis of Hi-C and expression data shows how 3D organization relates to gene expression programs, at baseline and under stress ([Nurick et al. 2018](https://doi.org/10.1186/s13072-018-0220-2)).

## Linking enhancers to genes
Computational models predict enhancer–gene links. [FOCS](https://doi.org/10.1186/s13059-018-1432-2) and [CT-FOCS](https://doi.org/10.1093/nar/gkac048) use correlated activity across samples. The [ABC model](https://doi.org/10.1038/s41588-019-0538-0) multiplies enhancer activity by contact frequency, and [EpiMap](https://doi.org/10.1038/s41586-020-03145-z) links enhancers across hundreds of tissues.

## Open questions
- Many genes have several redundant enhancers, which buffers them against mutations in one of them.
- Predicting enhancer–gene links in cell types without 3D data.
