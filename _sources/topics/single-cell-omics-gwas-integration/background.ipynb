# Background

## Motivation
Bulk tissue data average over many cell types, so they can hide the cells in which risk variants act. Single-cell RNA-seq and single-cell chromatin profiling (scATAC-seq) resolve cell types, cell states and developmental trajectories. Atlases such as the [Human Cell Atlas](https://doi.org/10.7554/eLife.27041) and [Tabula Sapiens](https://doi.org/10.1126/science.abl4896) provide these data across the body.

## Linking GWAS to cell types and states
- **Cell-type enrichment.** Genes specifically expressed in a cell type are tested for heritability enrichment ([LDSC-SEG](https://doi.org/10.1038/s41588-018-0081-4), [CELLECT](https://doi.org/10.7554/eLife.55851)). [sc-linker](https://doi.org/10.1038/s41588-022-01187-9) combines single-cell gene programs with enhancer–gene links.
- **Individual cells.** [scDRS](https://doi.org/10.1038/s41588-022-01167-z) scores each cell for the expression of disease genes, which shows variation within a cell type. [SCAVENGE](https://doi.org/10.1038/s41587-022-01341-y) propagates fine-mapped variants over a single-cell chromatin graph.
- **Developmental trajectories.** Integrating GWAS with single-cell data of developmental processes links differentiation trajectories to traits, for example the maturation of cortical inhibitory interneurons to schizophrenia, and prioritizes the genes behind each link ([Shulman et al. 2021](https://doi.org/10.1016/j.csbj.2021.05.055)).

## Single-cell QTLs
Single-cell eQTL studies, such as [OneK1K](https://doi.org/10.1126/science.abf3041), map genetic effects in each cell type of the blood. Many of these effects are invisible in bulk data. Methods include [SAIGE-QTL](https://doi.org/10.1038/s41588-026-02798-2), [CellRegMap](https://doi.org/10.15252/msb.202110663) for context-dependent effects and [tensorQTL](https://doi.org/10.1186/s13059-019-1836-7) for fast mapping. [SCENT](https://doi.org/10.1038/s41588-024-01682-1) links enhancers to genes from multimodal single-cell data, and [ArchR](https://doi.org/10.1038/s41588-021-00790-6) is a standard toolkit for scATAC-seq.

## Open questions
- Disease-relevant cell states may exist only in patients or after a stimulus.
- Single-cell QTL studies still have small sample sizes.
