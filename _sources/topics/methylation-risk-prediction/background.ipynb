# Background

## Why methylation
DNA methylation at CpG sites reflects both genetics and exposures such as smoking, diet and ageing, and it can be measured cheaply from blood on arrays. Methylation-based predictors can therefore capture risk that germline genetics misses ([Yousefi et al. 2022](https://doi.org/10.1038/s41576-022-00465-w)).

## Epigenetic clocks
- **First generation.** The [Horvath clock](https://doi.org/10.1186/gb-2013-14-10-r115) and the [Hannum clock](https://doi.org/10.1016/j.molcel.2012.10.016) predict chronological age from a few hundred CpGs.
- **Second generation.** These are trained on outcomes rather than age. [PhenoAge](https://doi.org/10.18632/aging.101414) is trained on clinical markers of ageing, and [GrimAge](https://doi.org/10.18632/aging.101684) on mortality and plasma protein surrogates; both predict lifespan and disease better than age clocks.
- **Pace of ageing.** [DunedinPACE](https://doi.org/10.7554/eLife.73420) measures the current rate of biological ageing.

## Methylation risk scores
Methylation risk scores (MRS) are built by EWAS-weighted sums or by penalized regression (elastic net) over CpGs. [EpiScores](https://doi.org/10.7554/eLife.71802) are methylation proxies for plasma proteins, and they improve prediction of several diseases.

## Open questions
- Reverse causation: methylation can be a consequence of disease rather than a cause.
- Cell-type composition of blood confounds methylation signals.
- How much MRS add on top of polygenic and clinical scores (see *Integrated risk prediction*).
