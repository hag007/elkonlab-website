# Background

## The problem
Most disease-associated variants are non-coding, and most non-coding variants have no measured effect. Sequence-to-function models learn the regulatory code from large functional genomics atlases, such as ENCODE and Roadmap. They can then predict the effect of any variant *in silico*, by comparing the prediction for the reference and the alternative allele.

## Generations of models
- **Convolutional models of chromatin profiles.** [DeepSEA](https://doi.org/10.1038/nmeth.3547) and [Basset](https://doi.org/10.1101/gr.200535.115) predict TF binding, DNase accessibility and histone marks from about 1 kb of sequence. [Sei](https://doi.org/10.1038/s41588-022-01102-2) groups thousands of these predictions into sequence classes.
- **Long-range models of expression.** [Basenji](https://doi.org/10.1101/gr.227819.117), [Enformer](https://doi.org/10.1038/s41592-021-01252-x) and [Borzoi](https://doi.org/10.1038/s41588-024-02053-6) predict RNA-seq coverage from 100–500 kb windows. [AlphaGenome](https://doi.org/10.1038/s41586-025-10014-0) predicts many modalities at base-pair resolution from 1 Mb of sequence.
- **Base-resolution models.** [ChromBPNet](https://doi.org/10.1101/2024.12.25.630221) models chromatin accessibility profiles at base resolution and corrects for the enzyme's sequence bias.
- **Genomic language models.** [Nucleotide Transformer](https://doi.org/10.1038/s41592-024-02523-z), [GPN](https://doi.org/10.1073/pnas.2311219120) and [Evo](https://doi.org/10.1126/science.ado9336) learn from unlabeled genomes and score variants by sequence likelihood, without functional labels.
- **Splicing.** [SpliceAI](https://doi.org/10.1016/j.cell.2018.12.015) and [Pangolin](https://doi.org/10.1186/s13059-022-02664-4) predict splice-site gain and loss and are widely used in clinical variant interpretation.
- **Integrative scores.** [CADD](https://doi.org/10.1038/ng.2892) combines many annotations into one deleteriousness score; [deltaSVM](https://doi.org/10.1038/ng.3331) is an early k-mer model of regulatory variant effects.

## Known limitations
Benchmarks show that current models capture promoter effects well but largely miss distal enhancers ([Karollus et al. 2023](https://doi.org/10.1186/s13059-023-02899-9)). They also explain little of the variation in gene expression between individuals, and often get the direction of effect wrong ([Huang et al. 2023](https://doi.org/10.1038/s41588-023-01574-w); [Sasse et al. 2023](https://doi.org/10.1038/s41588-023-01524-6)).

## Open questions
- Predicting effects in rare cell types and states that are missing from training data.
- Calibrating model scores so that they can serve as priors for fine-mapping and gene prioritization.
