# Background

## What a polygenic score is
A polygenic risk score (PRS) sums an individual's risk alleles across the genome, weighted by effect sizes estimated in a GWAS ([Torkamani et al. 2018](https://doi.org/10.1038/s41576-018-0018-x); [Choi et al. 2020](https://doi.org/10.1038/s41596-020-0353-1)). For several common diseases, people at the top of the PRS distribution have a risk similar to that of carriers of monogenic mutations ([Khera et al. 2018](https://doi.org/10.1038/s41588-018-0183-z)).

## Methods
- **Pruning and thresholding (P+T).** Keep independent SNPs below a p-value threshold, as in [PLINK](https://doi.org/10.1086/519795), [PLINK2](https://doi.org/10.1186/s13742-015-0047-8) and [PRSice-2](https://doi.org/10.1093/gigascience/giz082).
- **Bayesian shrinkage.** Model all SNPs jointly with an LD reference: [LDpred2](https://doi.org/10.1093/bioinformatics/btaa1029), [PRS-CS](https://doi.org/10.1038/s41467-019-09718-5) and [SBayesRC](https://doi.org/10.1038/s41588-024-01704-y).
- **Functional priors.** Weight SNPs by annotations: [LDpred-funct](https://doi.org/10.1038/s41467-021-25171-9), [PolyPred](https://doi.org/10.1038/s41588-022-01036-9) and SBayesRC.
- **Deep learning.** Model non-additive effects or use biological structure, for example [PRS-Net](https://doi.org/10.1101/2024.02.13.580211).

## Portability across ancestries
Most GWAS participants are of European ancestry, and PRS accuracy drops sharply in other populations ([Martin et al. 2019](https://doi.org/10.1038/s41588-019-0379-x)). The drop depends on the target population: European-based breast-cancer PRSs still identify Ashkenazi Jewish women in Israel at markedly elevated risk, with an odds ratio of about 1.6 per standard deviation ([Levi et al. 2023](https://doi.org/10.1136/jmg-2023-109185)). Multi-ancestry methods combine GWAS from several populations: [PRS-CSx](https://doi.org/10.1038/s41588-022-01054-7), [CT-SLEB](https://doi.org/10.1038/s41588-023-01501-z) and [TL-PRS](https://doi.org/10.1016/j.ajhg.2022.09.010). [PRANA](https://doi.org/10.64898/2026.07.12.26357860) takes a different route: a deep-learning model that adapts an existing European PRS to another ancestry using a modestly sized cohort from the target population, without a large GWAS there.

## Imputation and the target population
GWAS and PRS studies usually genotype people on SNP arrays and impute the untyped SNPs from a reference panel of sequenced individuals. Imputing a target population with a panel of its own ancestry usually improves imputation accuracy. However, when the GWAS and the target population differ in ancestry, an ancestry-matched imputation panel for the target set improves PRS prediction only marginally, and only under specific conditions ([Levi et al. 2024](https://doi.org/10.1093/bioinformatics/btae036)).
