# Background

## Why tools matter
Genomic methods have impact when other researchers can run them. Interactive tools and web servers make analyses accessible to experimental biologists, and workflow systems make computational analyses reproducible. Both follow the [FAIR principles](https://doi.org/10.1038/sdata.2016.18) of findable, accessible, interoperable and reusable research outputs.

## Interactive platforms and web servers
- **Expression analysis.** [EXPANDER](https://doi.org/10.1186/1471-2105-6-232) integrates clustering, biclustering, enrichment and promoter analysis of gene-expression data in one program ([Ulitsky et al. 2010](https://doi.org/10.1038/nprot.2009.230)). Its [current version](https://doi.org/10.1016/j.jmb.2019.05.013) supports RNA-seq and transcriptome analysis.
- **Pathways.** [SPIKE](https://doi.org/10.1186/1471-2105-9-110) is a curated database of signaling pathways with visualization and analysis tools ([Paz et al. 2011](https://doi.org/10.1093/nar/gkq1167)).
- **Networks and drug repurposing.** The [DOMINO web server](https://doi.org/10.1093/bioinformatics/btac067) runs active module identification from a browser. [Drugst.One](https://doi.org/10.1093/nar/gkae388) turns network tools into interactive web apps for drug repurposing.
- **General platforms.** [Galaxy](https://doi.org/10.1093/nar/gky379) runs analysis tools from the browser without programming.

## Workflows and reproducibility
- **Workflow managers.** [Snakemake](https://doi.org/10.1093/bioinformatics/bts480) and [Nextflow](https://doi.org/10.1038/nbt.3820) describe analyses as reproducible, portable pipelines.
- **Community pipelines.** [nf-core](https://doi.org/10.1038/s41587-020-0439-x) maintains peer-reviewed Nextflow pipelines for common analyses.
- **Software distribution.** [Bioconda](https://doi.org/10.1038/s41592-018-0046-7) packages thousands of bioinformatics tools for easy installation.

## Open questions
- Long-term maintenance of academic web servers after funding ends.
- Containers and workflow systems help, but reproducibility also needs versioned reference data.
