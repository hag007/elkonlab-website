# Background

## Motivation
Disease risk comes from several sources: rare high-penetrance variants (such as in BRCA1 or BRCA2), the polygenic background of common variants, family history, and clinical and lifestyle factors. Each alone gives a partial picture. Integrated models combine them into one estimate that is useful for screening and prevention ([Lewis & Vassos 2020](https://doi.org/10.1186/s13073-020-00742-5)).

## Common and rare variants together
Polygenic background modifies the penetrance of monogenic variants. Carriers of pathogenic variants in breast-cancer or cardiovascular genes have very different risks depending on their polygenic score ([Fahed et al. 2020](https://doi.org/10.1038/s41467-020-17374-3)). In BRCA1 and BRCA2 carriers, polygenic scores stratify breast and ovarian cancer risk ([Kuchenbaecker et al. 2017](https://doi.org/10.1093/jnci/djw302)). Rare-variant burden in genes found by sequencing studies (see *Rare variant analysis*) can add further information; for example, damaging missense variants in innate immunity genes are associated with earlier breast cancer onset in BRCA1 185delAG carriers ([Shemesh et al. 2026](https://doi.org/10.1136/jmg-2025-111394)).

## Clinical models
[BOADICEA](https://doi.org/10.1038/s41436-018-0406-9) combines pathogenic variants, a 313-SNP polygenic score ([Mavaddat et al. 2019](https://doi.org/10.1016/j.ajhg.2018.11.002)), family history and lifestyle factors to predict breast cancer risk. For cardiovascular disease, polygenic scores are tested on top of established clinical tools such as [QRISK3](https://doi.org/10.1136/bmj.j2099) and the [Pooled Cohort Equations](https://doi.org/10.1161/01.cir.0000437741.48606.98).

## Open questions
- How much polygenic scores add on top of good clinical models.
- Calibration across ancestries and across age.
- Combining genetic scores with methylation and other omic scores (see *Methylation risk prediction*).
