# Background

## Hereditary hearing loss
About 1 in 500 newborns has permanent hearing loss, and at least half of these cases are genetic ([Morton & Nance 2006](https://doi.org/10.1056/NEJMra050700)). Hereditary deafness is extremely heterogeneous, with over a hundred genes known. Mutations in GJB2 (connexin 26) are the most common cause in many populations ([Kelsell et al. 1997](https://doi.org/10.1038/387080a0)). Clinical sequencing panels find many novel variants, and the hearing-loss version of the [ACMG/AMP guidelines](https://doi.org/10.1002/humu.23630) and the [Deafness Variation Database](https://doi.org/10.1016/j.ajhg.2018.08.006) are used to classify them.

## Age-related hearing loss
Age-related hearing loss (ARHL) is common and complex. GWAS in large biobanks such as [UK Biobank](https://doi.org/10.1038/s41586-018-0579-z) and the [Million Veteran Program](https://doi.org/10.1016/j.jclinepi.2015.09.016) have identified dozens of loci ([Wells et al. 2019](https://doi.org/10.1016/j.ajhg.2019.09.008); [Trpchevska et al. 2022](https://doi.org/10.1016/j.ajhg.2022.04.010)). Studies that integrated these loci with cochlear expression data disagreed on the key cell types: some pointed to the stria vascularis ([Trpchevska et al. 2022](https://doi.org/10.1016/j.ajhg.2022.04.010)). An analysis of four ARHL GWAS datasets with four independent inner-ear single-cell datasets points instead to the cochlear sensory epithelial cells, and explains why earlier analyses missed this signal ([Eshel et al. 2024](https://doi.org/10.1016/j.ajhg.2024.01.008)).

## Inner-ear genomics
Interpreting hearing-loss genes needs data on the inner ear's many cell types ([Hertzano & Elkon 2012](https://doi.org/10.1016/j.heares.2012.01.002)).
- **Cell-type-specific transcriptomes** of sorted inner-ear cells showed key roles for Zeb1 and miR-200b ([Hertzano et al. 2011](https://doi.org/10.1371/journal.pgen.1002309)) and identified RFX transcription factors as essential for hair-cell maturation and hearing ([Elkon et al. 2015](https://doi.org/10.1038/ncomms9549)).
- **Hair-cell regulators.** Helios controls outer hair cell maturation ([Chessum et al. 2018](https://doi.org/10.1038/s41586-018-0728-4)), and GFI1 represses neuronal genes in developing hair cells ([Matern et al. 2020](https://doi.org/10.1242/dev.186015)).
- **Response to noise.** A cell-type-specific atlas describes the inner ear's response to acoustic trauma ([Milon et al. 2021](https://doi.org/10.1016/j.celrep.2021.109758)).
- **Shared resources.** The [gEAR portal](https://doi.org/10.1038/s41592-021-01200-9) collects inner-ear expression data for the community.

## Open questions
- Most heritability of ARHL is still unexplained.
- Translating genetic findings into therapies, such as hair-cell regeneration ([Khalaily et al. 2026](https://doi.org/10.1126/sciadv.aed3887)).
