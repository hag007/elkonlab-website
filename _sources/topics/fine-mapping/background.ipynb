# Background

## The problem
A GWAS locus usually contains tens to hundreds of variants in strong linkage disequilibrium (LD) with the lead SNP. Fine-mapping asks which of them are causal. It usually reports a *posterior inclusion probability* (PIP) for each variant and *credible sets* that contain a causal variant with high probability ([Schaid et al. 2018](https://doi.org/10.1038/s41576-018-0016-z)).

## Statistical methods
- **Bayesian fine-mapping.** Early methods assumed a single causal variant per locus ([Maller et al. 2012](https://doi.org/10.1038/ng.2435)). [FINEMAP](https://doi.org/10.1093/bioinformatics/btw018) searches over configurations with several causal variants. [SuSiE](https://doi.org/10.1111/rssb.12388) writes the signal as a sum of single effects, which gives one credible set per effect. [SuSiE-RSS](https://doi.org/10.1371/journal.pgen.1010299) runs it on summary statistics.
- **Robustness to LD mismatch.** Summary-statistics fine-mapping needs an LD matrix, often from an external reference panel. Mismatch between this panel and the GWAS creates false signals. [SLALOM](https://doi.org/10.1016/j.xgen.2022.100210) and [DENTIST](https://doi.org/10.1038/s41467-021-27438-7) detect problematic loci or variants, and [CARMA](https://doi.org/10.1038/s41588-023-01392-0) models the mismatch directly.

## Functional priors
Causal variants are enriched in functional annotations. [PAINTOR](https://doi.org/10.1371/journal.pgen.1004722) and [PolyFun](https://doi.org/10.1038/s41588-020-00735-5) learn annotation-based priors and use them to sharpen PIPs. Scores from sequence models (see *Non-coding variant effect prediction*) are a natural source for such priors.

## Multiple ancestries
LD differs between populations, so combining ancestries narrows credible sets. Methods include [MsCAVIAR](https://doi.org/10.1371/journal.pgen.1009733), [SuSiEx](https://doi.org/10.1038/s41588-024-01870-z) and [MESuSiE](https://doi.org/10.1038/s41588-023-01604-7).

## Open questions
- Causal variants that are missing from the data (not genotyped or imputed) can produce confident false signals.
- How to validate PIPs at scale, for example with MPRA or CRISPR screens.
